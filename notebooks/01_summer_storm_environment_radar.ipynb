## 📖 1. แนวคิดพื้นฐาน

### 1.1 พายุฤดูร้อนในภาคอีสาน
ช่วง **ก่อนมรสุม (pre-monsoon) มีนาคม–พฤษภาคม** ประเทศไทยตอนบนร้อนจัด อากาศใกล้ผิวพื้นร้อนและชื้น
เมื่อมี **ลิ่มความกดอากาศสูงจากจีน** แผ่ลงมา หรือมีคลื่นในลมชั้นบนผ่าน อากาศร้อนชื้นจะถูกยกตัวจนเกิด **พายุฝนฟ้าคะนอง ลมกระโชกแรง และลูกเห็บ**
ภูมิประเทศก็มีบทบาท: ที่ราบสูงและภูเขาด้านตะวันตกของอีสาน (เช่น **ภูกระดึง สูง ~1,300 m**) ร้อนเร็วตอนกลางวันและมักเป็นจุดที่พายุเริ่มก่อตัว

### 1.2 ดัชนีจากข้อมูลตรวจอากาศชั้นบน (sounding)
| ดัชนี | ความหมาย | ค่าที่ควรรู้ |
|---|---|---|
| **CAPE** | พลังงานลอยตัวของก้อนอากาศที่ยกตัวขึ้น | > 2,500 J/kg = ไม่มีเสถียรภาพมาก |
| **CIN** | พลังงานที่ต้องเอาชนะก่อนก้อนอากาศจะลอยได้เอง | ยิ่งติดลบมาก ยิ่งเกิดพายุยาก |
| **DCAPE** | พลังงานของกระแสอากาศจมจากการระเหย (downdraft) | > 1,000 J/kg = เสี่ยง **downburst / ลมกระโชกแรง** |
| **ระดับเยือกแข็ง (0 °C)** | ความสูงที่น้ำเริ่มเป็นน้ำแข็ง | ยิ่งต่ำ ลูกเห็บยิ่งละลายไม่ทันก่อนถึงพื้น |
| **Wind shear 0–6 km** | ลมเปลี่ยนตามความสูงมากแค่ไหน | < 10 m/s = พายุเซลล์เดี่ยว/หลายเซลล์อายุสั้น; > 20 m/s = supercell |
| **ลมเฉลี่ย 850–300 hPa** | "steering wind" ที่พัดพาเซลล์พายุ | ใช้เทียบกับการเคลื่อนที่ที่เห็นในเรดาร์ (NB3) |

$$
\mathrm{CAPE} = g\int_{LFC}^{EL} \frac{T_{v,\mathrm{parcel}} - T_{v,\mathrm{env}}}{T_{v,\mathrm{env}}}\,dz
\qquad
\mathrm{DCAPE} = g\int_{z_{\mathrm{sfc}}}^{z_{d}} \frac{T_{v,\mathrm{env}} - T_{v,\mathrm{down}}}{T_{v,\mathrm{env}}}\,dz
$$

### 1.3 การตีความค่า reflectivity
$$ \mathrm{dBZ} = 10\log_{10} Z,\qquad Z\ [\mathrm{mm^6\,m^{-3}}] $$
| dBZ | ลักษณะโดยทั่วไป |
|---|---|
| 15–30 | ฝนเบา / ฝนแบบ stratiform |
| 30–40 | ฝนปานกลาง |
| 40–50 | ฝนหนัก แกนพายุพาความร้อน |
| **≥ 50–55** | ฝนหนักมาก และ **อาจมีลูกเห็บ** (ลูกเห็บสะท้อนคลื่นแรงกว่าหยดน้ำมาก) |

### 1.4 ข้อมูลเรดาร์ KKN ในรายวิชานี้: pseudo-CAPPI 2 km
เรดาร์ขอนแก่น (C-band, กรมอุตุนิยมวิทยา) ใน volume นี้สแกน **เพียง 4 มุมยก (0.48°, 1.48°, 2.39°, 3.39°)**
notebook `00_prepare_pcappi_for_pysteps` สร้าง **pseudo-CAPPI 2 km** (ระดับความสูงคงที่ 2 km เหนือเรดาร์) บนกริด 1 km รัศมี 240 km ทุก 15 นาที
- ระยะ **~32–130 km**: ได้ค่าที่ 2 km จริง (ประมาณค่าระหว่างมุมยก)
- **ใกล้กว่า ~32 km**: ใช้มุมยกบนสุด (ลำคลื่นต่ำกว่า 2 km)
- **ไกลกว่า ~130 km**: ใช้มุมยกต่ำสุด ซึ่ง **สูง 2–5 km** → ค่าอาจต่ำกว่าความจริง

**อ่านรูปนี้อย่างไร?**

## 📈 4. ภาพรวมทั้งวันจากเรดาร์: วงจรชีวิตของกลุ่มฝน

```python
area = kt.echo_area(Z, thresholds=(20, 30, 40, 50))
date0 = ts[0].replace(hour=0, minute=0)
fig, ax = plt.subplots(figsize=(8.5, 3.0), layout="constrained")
for thr, c in zip((20, 30, 40, 50), ["#9ecae1", "#4292c6", "#08519c", "#08306b"]):
    a = np.where(meta["available"], area[thr], np.nan)
    ax.plot(ts, a / 1e3, color=c, lw=1.2, marker="o", ms=1.5, label=f"≥ {thr} dBZ")
for t, ok in zip(ts, meta["available"]):
    if not ok:
        ax.axvspan(t - pd.Timedelta(minutes=7.5), t + pd.Timedelta(minutes=7.5), color="0.85", lw=0)
phases = [("05:00", "06:00", "initiation"), ("06:00", "10:00", "growth & peak"),
          ("10:00", "13:30", "upscale growth"), ("13:30", "16:30", "decay")]
for a0, a1, lab in phases:
    t0 = date0.replace(hour=int(a0[:2]), minute=int(a0[3:]))
    t1 = date0.replace(hour=int(a1[:2]), minute=int(a1[3:]))
    ax.axvspan(t0, t1, color="#fff3cd", alpha=0.5, lw=0, zorder=0)
    ax.text(t0 + (t1 - t0) / 2, 9.6, lab, ha="center", fontsize=6.5, color="0.3")
for _, r in reports.drop_duplicates("time_UTC").iterrows():
    t = kt.report_time_utc(r, date0)
    if t is not None:
        ax.axvline(t, color="#d7191c", lw=0.8, ls="--")
        ax.text(t, 8.6, r["province"], rotation=90, fontsize=6, color="#d7191c", ha="right", va="top")
ax.set_ylim(0, 10)
ax.set_xlim(ts[0], ts[-1])
mp.time_axis(ax)
mp.skill_axes(ax, xlabel="Time (UTC)", ylabel="Echo area (10$^3$ km$^2$)")
ax.legend(loc="upper right", ncol=2, fontsize=6.5, frameon=True, framealpha=0.9)
mp.save_figure(fig, "nb1_echo_area_timeseries")
plt.show()
```

| **รวมตัวเป็นระบบใหญ่** | 10:00–13:30 (17:00–20:30) | พื้นที่ ≥ 20 dBZ ยังเพิ่ม (สูงสุด ~9,000 km²) แต่ ≥ 40 dBZ ลดลง = ฝน stratiform ขยายตัว |

```python
hours = [f"{h:02d}:00" for h in range(5, 17)]           # 12:00–23:00 น.
fig, axs = mp.map_panels(3, 4, meta, panel_width=2.3)
for k, (ax, hh) in enumerate(zip(axs, hours)):
    i = time_index(meta, hh)
    im = mp.radar_map(ax, Z[i], meta, title=mp.tlabel(ts[i]), scalebar=(k == 8), north=(k == 3),
                      gridline_labels=(("left",) if k % 4 == 0 else ()) + (("bottom",) if k >= 8 else ()))
mp.panel_labels(axs)
mp.add_colorbar(fig, im, axs, shrink=0.5)
mp.save_figure(fig, "nb1_hourly_maps")
plt.show()
```

## 📍 6. อนุกรมเวลา reflectivity ณ จุดที่เกิดความเสียหาย
ใช้ค่า **สูงสุดในรัศมี 5 km** รอบ centroid ของตำบล (เพราะตำแหน่งความเสียหายจริงไม่ใช่จุดเดียว)

```python
names = list(kt.KEY_SITES)
fig, axs = plt.subplots(2, 3, figsize=(8.8, 4.6), sharex=True, sharey=True, layout="constrained")
for ax, name in zip(axs.ravel(), names):
    r = kt.site(reports, name)
    s = kt.point_series(Z, meta, r["lon"], r["lat"], radius_km=5)
    s = np.where(meta["available"], s, np.nan)
    ax.plot(ts, s, color="0.2", lw=1.1)
    ax.axhspan(50, 70, color="#f4a582", alpha=0.3, lw=0)
    t_rep = kt.report_time_utc(r, date0)
    if t_rep is not None:
        ax.axvline(t_rep, color="#d7191c", ls="--", lw=1)
    ax.set_title(f"{name} ({r['range_km']:.0f} km)", loc="left", fontsize=8)
    ax.set_ylim(0, 60)
    ax.set_xlim(date0.replace(hour=5), date0.replace(hour=17))
    mp.skill_axes(ax, xlabel="", ylabel="Max $Z_H$ (dBZ)" if ax in axs[:, 0] else "")
for ax in axs[1]:
    mp.time_axis(ax, local_top=False, interval_h=3)
mp.save_figure(fig, "nb1_site_timeseries")
plt.show()
```

เส้นประแดง = เวลาที่รายงาน, แถบส้ม = ≥ 50 dBZ (ฝนหนักมาก/อาจมีลูกเห็บ)
- **ภูกระดึงและนากลาง/สุวรรณคูหา:** ค่าสูงสุดตรงกับเวลารายงานภายใน ~15 นาที → เรดาร์ "ยืนยัน" รายงานได้
- **ศรีเชียงใหม่ (หนองคาย):** ค่าสูงสุดมาหลังเวลาในข่าว 30–60 นาที และค่าต่ำกว่า (~45 dBZ) — ทำไม? (แบบฝึกหัด 1.3)
- **บึงกาฬ / ขอนแก่น:** ข่าวระบุแค่ "กลางคืน" → เรดาร์ช่วยบอกเวลาได้ (ดู NB6)

### 🧪 แบบฝึกหัด 1.2: พายุลูกแรกเกิดที่ไหน?
หา **เวลาแรก หลัง 05:00 UTC** ที่มี pixel ≥ 35 dBZ ภายในรัศมี 160 km แล้วหา **ตำแหน่ง (lon, lat)** ของ pixel ที่มีค่าสูงสุดในเวลานั้น
จากนั้นหา **ความสูงภูมิประเทศ** ที่จุดนั้นด้วย `mp.terrain_at(meta, lon, lat)` และ **ระยะห่างจากภูกระดึง**

**คำถาม:** จุดเริ่มต้นของพายุสัมพันธ์กับภูมิประเทศอย่างไร?

<details>
<summary><b>💡 คำใบ้</b></summary>

- `mask = kt.range_mask(meta, 160)`
- วน `for i in range(time_index(meta, "05:00"), len(ts)):` แล้วเช็ก `np.nanmax(np.where(mask, Z[i], np.nan)) >= 35`
- ตำแหน่ง: `r, c = np.unravel_index(np.nanargmax(np.where(mask, Z[i], np.nan)), Z[i].shape)` แล้ว `meta["lon"][r, c]`, `meta["lat"][r, c]`
- ระยะ (km): `np.hypot(x1 - x2, y1 - y2) / 1000` โดยใช้ `kt.lonlat_to_xy`

</details>

```python
# ✍️ แบบฝึกหัด 1.2 — เขียนโค้ดของคุณที่นี่
# mask = kt.range_mask(meta, 160)
# for i in range(time_index(meta, "05:00"), len(ts)):
#     ...
```

```python
#@title ✅ เฉลยแบบฝึกหัด 1.2
mask = kt.range_mask(meta, 160)
for i in range(time_index(meta, "05:00"), len(ts)):
    zi = np.where(mask, Z[i], np.nan)
    if np.nanmax(zi) >= 35:
        break
r_, c_ = np.unravel_index(np.nanargmax(zi), zi.shape)
lon0, lat0 = meta["lon"][r_, c_], meta["lat"][r_, c_]
pk = kt.site(reports, "Phu Kradueng")
xa, ya = kt.lonlat_to_xy(meta, lon0, lat0)
dist = np.hypot(xa - pk["x"], ya - pk["y"]) / 1000
print(f"echo ≥ 35 dBZ ครั้งแรก: {mp.tlabel(ts[i])}  ค่าสูงสุด {np.nanmax(zi):.1f} dBZ")
print(f"ตำแหน่ง {lat0:.2f}°N {lon0:.2f}°E | ภูมิประเทศสูง {mp.terrain_at(meta, lon0, lat0):.0f} m | ห่างภูกระดึง {dist:.0f} km")

fig, axs = mp.map_panels(1, 1, meta, panel_width=4.0)
ax = axs[0]
zoom = kt.box_around(meta, lon0, lat0, 80)
ax.set_extent(zoom, crs=mp.get_crs(meta))
mp.add_terrain(ax, meta)
mp.add_basemap(ax, meta, province_labels=True)
mp.obs_contour(ax, Z[i], meta, level=35, color="#d7191c", linewidth=1.0, label="35 dBZ at first echo")
mp.add_reports(ax, reports, legend=False)
ax.set_title(f"First ≥35 dBZ echo, {mp.tlabel(ts[i])}", loc="left", fontsize=8)
mp.add_scalebar(ax)
mp.save_figure(fig, "nb1_ex12_first_echo")
plt.show()
```

### 🧪 แบบฝึกหัด 1.3: เรดาร์กับรายงานความเสียหายห่างกันกี่นาที?
สำหรับจุด `Phu Kradueng`, `Na Klang`, `Suwannakhuha`, `Si Chiang Mai` ให้หา
1. เวลาที่ค่าสูงสุดในรัศมี 5 km **มากที่สุด** ในช่วง ±2 ชั่วโมงรอบเวลารายงาน
2. **ผลต่าง (นาที)** = เวลาที่เรดาร์สูงสุด − เวลารายงาน
3. ค่า dBZ สูงสุด ระยะจากเรดาร์ และความสูงของข้อมูล (`data_height_m`)

**คำถาม:** ทำไมจุดที่อยู่ไกลจากเรดาร์ (หนองคาย) จึงมีค่าต่ำกว่าและเวลาคลาดเคลื่อนมากกว่า? (คิดถึง: ลำคลื่นสูงขึ้นและกว้างขึ้นตามระยะ, การลดทอนสัญญาณ *attenuation* ของคลื่น C-band ผ่านฝนหนัก, และความแม่นยำของเวลาในข่าว)

<details>
<summary><b>💡 คำใบ้</b></summary>

- `t_rep = kt.report_time_utc(r, date0)` และ `win = [i for i, t in enumerate(ts) if abs((t - t_rep).total_seconds()) <= 7200]`
- `s = kt.point_series(Z, meta, r["lon"], r["lat"], 5)` แล้วหา `i_max = win[np.nanargmax(s[win])]`
- ผลต่าง `(ts[i_max] - t_rep).total_seconds() / 60`

</details>

```python
# ✍️ แบบฝึกหัด 1.3 — เขียนโค้ดของคุณที่นี่
# rows = []
# for name in ["Phu Kradueng", "Na Klang", "Suwannakhuha", "Si Chiang Mai"]:
#     r = kt.site(reports, name)
#     ...
```

```python
#@title ✅ เฉลยแบบฝึกหัด 1.3
rows = []
for name in ["Phu Kradueng", "Na Klang", "Suwannakhuha", "Si Chiang Mai"]:
    r = reports.loc[reports["id"] == kt.KEY_SITES[name]].iloc[0]
    t_rep = kt.report_time_utc(r, date0)
    s = np.where(meta["available"], kt.point_series(Z, meta, r["lon"], r["lat"], 5), np.nan)
    win = [i for i, t in enumerate(ts) if abs((t - t_rep).total_seconds()) <= 7200]
    i_max = win[int(np.nanargmax(s[win]))]
    rows.append({"site": name, "report_LT": r["time_LT"], "radar_peak_LT": kt.hhmm(ts[i_max], local=True),
                 "lag_min": (ts[i_max] - t_rep).total_seconds() / 60, "max_dBZ": s[i_max],
                 "range_km": r["range_km"], "data_height_m": r["data_height_m"]})
lag = pd.DataFrame(rows).round(1)
print(lag.to_string(index=False))
print("\n→ ยิ่งไกล ลำคลื่นยิ่งสูงและกว้าง (ที่ 165 km กว้าง ~2.9 km) ค่าเฉลี่ยในลำคลื่นจึงต่ำลง และคลื่น C-band ถูกลดทอนเมื่อผ่านแกนฝนหนัก"
      "\n  นอกจากนี้เวลาในข่าวมักเป็นค่าประมาณ ('ประมาณ 17.00 น.') ส่วนรายงานของ ปภ. (ภูกระดึง) ระบุเวลาชัดเจนกว่า")
```
